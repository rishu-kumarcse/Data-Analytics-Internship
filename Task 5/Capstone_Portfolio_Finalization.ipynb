````python
def generate_readme():
    """
    Generates a polished README.md for the master portfolio repository.
    Serves as the project homepage with overview, links, skills, and learnings.
    """

    # Build project links section
    repo_section = ""
    for i, (title, link) in enumerate(REPO_LINKS.items(), 1):
        repo_section += f"| {i} | [{title}]({link}) | [🔗 View Repository]({link}) |\n"

    # Build skills section
    skills_section = "\n".join([f"- ✅ {skill}" for skill in SKILLS])

    # Build key learnings section
    learnings_section = "\n".join([f"- 💡 {learning}" for learning in KEY_LEARNINGS])

    readme_content = f"""# 🎓 {MASTER_REPO_NAME}

![Portfolio Banner](https://img.shields.io/badge/Internship-Portfolio-blue?style=for-the-badge)
![Role](https://img.shields.io/badge/Role-{INTERNSHIP_ROLE.replace(' ', '%20')}-green?style=for-the-badge)
![Status](https://img.shields.io/badge/Status-Completed-brightgreen?style=for-the-badge)

---

## 👤 About

| Field | Details |
|---|---|
| **Name** | {YOUR_NAME} |
| **Role** | {INTERNSHIP_ROLE} |
| **Company** | {INTERNSHIP_COMPANY} |
| **Duration** | {INTERNSHIP_DURATION} |
| **LinkedIn** | [{YOUR_NAME}]({LINKEDIN_PROFILE}) |
| **GitHub** | [@{GITHUB_USERNAME}](https://github.com/{GITHUB_USERNAME}) |

---

## 📌 Overview

This repository is the **master portfolio** for my {INTERNSHIP_ROLE} internship at **{INTERNSHIP_COMPANY}**.
It consolidates all four project phases into a single, polished reference point — demonstrating
a complete and masterful end-to-end data analytics project.

---

## 📁 Project Repositories

| # | Project Phase | Repository Link |
|---|---|---|
{repo_section}
---

## 🛠️ Technical Skills Demonstrated

{skills_section}

---

## 💡 Key Learnings

{learnings_section}

---

## 📊 Project Structure

```
{MASTER_REPO_NAME}/
│
├── README.md                  ← You are here (Portfolio Homepage)
├── presentation/
│   └── final_presentation.pdf ← Refined presentation deck
├── reports/
│   └── summary_report.md      ← Key learnings & technical summary
└── links.md                   ← All repository links
```

---

## 📬 Connect With Me

- 🔗 LinkedIn: [{YOUR_NAME}]({LINKEDIN_PROFILE})
- 💻 GitHub: [@{GITHUB_USERNAME}](https://github.com/{GITHUB_USERNAME})

---

*This portfolio was created as part of the Capstone Integration & Portfolio Finalization phase of the internship.*
"""
    return readme_content


readme = generate_readme()

# Save to file
with open("README.md", "w", encoding="utf-8") as f:
    f.write(readme)

print("✅ README.md generated successfully!")
print("\n" + "="*60)
print(readme[:800] + "\n... [truncated for preview]")
````

## ✅ Step 5: Final Review Checklist

````python
def run_final_checklist():
    """
    Validates that all required portfolio components have been generated
    and prints a final checklist for the user.
    """
    import os

    checklist = {
        "README.md generated": os.path.exists("README.md"),
        "LinkedIn post generated": os.path.exists("linkedin_post.txt"),
        "GitHub Pages HTML generated": os.path.exists("index.html"),
        "Your name configured": YOUR_NAME != "YourName",
        "GitHub username configured": GITHUB_USERNAME != "your-github-username",
        "All 4 repo links added": all("your-username" not in v for v in REPO_LINKS.values()),
        "Skills list filled": len(SKILLS) >= 3,
        "Key learnings filled": len(KEY_LEARNINGS) >= 2,
    }

    print("=" * 55)
    print("   📋 FINAL PORTFOLIO REVIEW CHECKLIST")
    print("=" * 55)

    all_pass = True
    for item, status in checklist.items():
        icon = "✅" if status else "❌"
        print(f"  {icon}  {item}")
        if not status:
            all_pass = False

    print("=" * 55)

    if all_pass:
        print("\n🎉 ALL CHECKS PASSED! Your portfolio is ready.")
        print(f"\n📁 Master Repo Name : {MASTER_REPO_NAME}")
        print(f"🌐 GitHub Pages URL  : https://{GITHUB_USERNAME}.github.io/{MASTER_REPO_NAME}/")
        print(f"💼 LinkedIn Post     : linkedin_post.txt")
    else:
        print("\n⚠️  Some items need attention. Please fix the ❌ items above.")
        print("   Go back to Step 1 and update your configuration.")


run_final_checklist()
````
